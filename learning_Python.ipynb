## 1. Print & comments
- `print()` shows output. `sep` changes the separator, `end` changes what is printed at the end.
- `#` starts a single-line comment. Python has no real multi-line comment; a triple-quoted string is used for docstrings.

```python
# This is a comment
print("Hello, World!")
print("a", "b", "c")                # a b c
print("a", "b", "c", sep="-")        # a-b-c
print("no newline", end=" | ")
print("same line")

print("Tab:\tafter tab")
print("New\nline")
print("Quote: \"hi\" and 'hi'")
print("Backslash: \\")
print(r"Raw string keeps \n as it is")
```

## 2. Variables & naming
- No type declaration needed; Python is **dynamically typed**.
- Rules: start with a letter or `_`, no spaces, case-sensitive, cannot be a keyword (`if`, `for`, `class`...).
- Convention: `snake_case` for variables/functions, `CamelCase` for classes, `UPPER_CASE` for constants.

```python
x = 10
name = "Aman"
pi = 3.14
is_ready = True

# multiple assignment
a, b, c = 1, 2, 3
p = q = r = 0            # same value to all

# swap without temp
a, b = b, a
print(a, b)              # 2 1

# the type can change later (dynamic typing)
x = "now a string"
print(x, type(x))

# useful: see all keywords
import keyword
print(keyword.kwlist)
```

## 6. Operators

```python
a, b = 17, 5

# Arithmetic
print(a + b, a - b, a * b)
print(a / b)       # 3.4  true division (always float)
print(a // b)      # 3    floor division
print(a % b)       # 2    remainder
print(a ** b)      # 1419857  power
print(-17 // 5)    # -4   floor goes DOWN, not toward zero
print(-17 % 5)     # 3    result has the sign of the divisor
print(divmod(a, b))   # (3, 2)

# Comparison (can be chained)
print(1 < 5 < 10)      # True
print(a == b, a != b, a >= b)

# Logical
print(True and False, True or False, not True)
print(0 or "default")  # 'default'  (or returns first truthy value)
print(5 and 7)         # 7          (and returns last value if all truthy)

# Assignment
n = 10
n += 5; n -= 3; n *= 2; n //= 5; n **= 2
print(n)

# Identity and membership
l1 = [1, 2]; l2 = [1, 2]; l3 = l1
print(l1 == l2, l1 is l2, l1 is l3)   # True False True
print(3 in [1, 2, 3], "a" not in "cat")  # True False

# Bitwise
print(5 & 3, 5 | 3, 5 ^ 3, ~5, 5 << 1, 5 >> 1)   # 1 7 6 -6 10 2
```

## 7. Strings
Strings are **immutable** ordered sequences of characters. Indexing starts at 0; negative indexes count from the end.

```python
s = "Python Programming"
print(len(s))
print(s[0], s[-1])        # P g
print(s[0:6])             # Python       (stop index is excluded)
print(s[7:])              # Programming
print(s[:6])              # Python
print(s[::2])             # every 2nd char
print(s[::-1])            # reversed string
# s[0] = "J"  -> TypeError (immutable)
```

```python
t = "  Hello, World  "
print(t.strip(), "|", t.lstrip(), "|", t.rstrip())
print(t.upper(), t.lower(), t.title(), t.swapcase())
print(t.replace("World", "Python"))
print(t.count("l"), t.find("o"), t.find("z"))   # find returns -1 if missing
print("Hello".index("l"))                       # index raises ValueError if missing
print("a,b,c".split(","))                       # ['a', 'b', 'c']
print("-".join(["a", "b", "c"]))                 # a-b-c
print("hello".startswith("he"), "hello".endswith("lo"))
print("abc123".isalnum(), "abc".isalpha(), "123".isdigit(), "  ".isspace())
print("hello".capitalize(), "7".zfill(3), "hi".center(10, "*"))
```

```python
name, marks, pct = "Riya", 87, 87.456

print("Name: %s, Marks: %d" % (name, marks))        # old style
print("Name: {}, Marks: {}".format(name, marks))    # .format
print(f"Name: {name}, Marks: {marks}")              # f-string (best)

print(f"{pct:.2f}")        # 87.46   2 decimal places
print(f"{marks:05d}")      # 00087   zero padded
print(f"{marks:>8}|")      # right aligned in width 8
print(f"{name:<8}|")       # left aligned
print(f"{1234567:,}")      # 1,234,567
print(f"{0.256:.1%}")      # 25.6%
print(f"{marks=}")         # marks=87  (debug style, Python 3.8+)

# multi-line string
poem = '''Line one
Line two'''
print(poem)
```

## 8. Lists
Ordered, **mutable**, allow duplicates, can hold mixed types.

```python
nums = [10, 20, 30, 40, 50]
print(nums[0], nums[-1], nums[1:4], nums[::-1])

nums.append(60)            # add at end
nums.insert(1, 15)         # insert at index
nums.extend([70, 80])      # add many
print(nums)

nums.remove(15)            # remove first occurrence of VALUE
popped = nums.pop()        # remove and return last (pop(i) for index i)
del nums[0]                # delete by index
print(nums, popped)

print(len(nums), nums.index(30), nums.count(30), 30 in nums)

data = [5, 2, 9, 1]
data.sort()                # in place, returns None
print(data)
data.sort(reverse=True)
print(data)
print(sorted([3, 1, 2]))   # returns a NEW list
data.reverse()
print(data)
```

```python
# Copying: assignment does NOT copy
a = [1, 2, 3]
b = a            # same list
c = a.copy()     # new list (also a[:] or list(a))
a.append(4)
print(b, c)      # [1, 2, 3, 4] [1, 2, 3]

# Nested lists (2D) and the classic trap
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
print(matrix[1][2])                       # 6
good = [[0] * 3 for _ in range(2)]        # independent rows
bad = [[0] * 3] * 2                       # SAME row repeated
good[0][0] = 1
bad[0][0] = 1
print(good)      # [[1, 0, 0], [0, 0, 0]]
print(bad)       # [[1, 0, 0], [1, 0, 0]]

# Unpacking
first, *middle, last = [1, 2, 3, 4, 5]
print(first, middle, last)    # 1 [2, 3, 4] 5

# Concatenate and repeat
print([1, 2] + [3], [0] * 4)
```

## 9. Tuples
Ordered, **immutable**, allow duplicates. Faster than lists and usable as dict keys.

```python
t = (1, 2, 3, 2)
print(t[0], t[-1], t[1:3], t.count(2), t.index(3))

single = (5,)        # comma makes it a tuple
not_tuple = (5)      # just an int
print(type(single), type(not_tuple))

x, y = (10, 20)      # unpacking
print(x, y)

# t[0] = 9 -> TypeError
# but a mutable item inside can change:
m = ([1, 2], 3)
m[0].append(99)
print(m)             # ([1, 2, 99], 3)
```

## 10. Sets
Unordered, **unique** elements, mutable. Very fast membership tests.

```python
s = {1, 2, 3, 3, 2}
print(s)                    # {1, 2, 3}
empty_set = set()           # {} would create an empty DICT
print(type({}), type(empty_set))

s.add(4); s.discard(10)     # discard: no error if missing
s.remove(4)                 # remove: KeyError if missing
print(s)

A = {1, 2, 3, 4}
B = {3, 4, 5, 6}
print(A | B)    # union         {1, 2, 3, 4, 5, 6}
print(A & B)    # intersection  {3, 4}
print(A - B)    # difference    {1, 2}
print(A ^ B)    # symmetric diff {1, 2, 5, 6}
print({1, 2} <= A, A.issuperset({1, 2}))

# Common use: remove duplicates
print(list(set([1, 2, 2, 3, 3, 3])))
```

## 11. Dictionaries
Key → value pairs. Keys must be immutable and unique. Insertion order is preserved (Python 3.7+).

```python
student = {"name": "Neha", "age": 20, "marks": 91}

print(student["name"])
print(student.get("city"))            # None (no error)
print(student.get("city", "Delhi"))   # default value

student["city"] = "Pune"              # add
student["age"] = 21                   # update
student.update({"marks": 95, "grade": "A"})
removed = student.pop("grade")        # remove key, return value
del student["city"]
print(student, removed)

print(student.keys())
print(student.values())
print(student.items())

for key, value in student.items():
    print(key, "->", value)

print("name" in student, "xyz" in student)   # membership checks KEYS

# Counting pattern (very common in exams)
text = "banana"
freq = {}
for ch in text:
    freq[ch] = freq.get(ch, 0) + 1
print(freq)          # {'b': 1, 'a': 3, 'n': 2}

# Sort a dict by value
print(sorted(freq.items(), key=lambda kv: kv[1], reverse=True))

# Dict comprehension
print({n: n ** 2 for n in range(1, 6)})
```

## 12. Conditionals
Indentation (4 spaces) defines blocks. Colon `:` after every condition.

```python
marks = 72

if marks >= 90:
    grade = "A"
elif marks >= 75:
    grade = "B"
elif marks >= 60:
    grade = "C"
else:
    grade = "F"
print(grade)                  # C

# Ternary (one-line if/else)
status = "pass" if marks >= 40 else "fail"
print(status)

# Truthy / falsy values
items = []
if not items:
    print("list is empty")

# Nested + combined conditions
age, has_id = 19, True
if age >= 18 and has_id:
    print("allowed")

# match-case (Python 3.10+)
day = 3
match day:
    case 1: print("Mon")
    case 2: print("Tue")
    case 3: print("Wed")
    case _: print("Other")
```

## 14. Functions

```python
def add(a, b):
    '''Return the sum of a and b.'''
    return a + b

print(add(2, 3))
print(add.__doc__)

# Default and keyword arguments
def greet(name, greeting="Hello"):
    return f"{greeting}, {name}!"

print(greet("Sam"))
print(greet("Sam", greeting="Hi"))
print(greet(greeting="Yo", name="Ann"))     # order does not matter with keywords

# Returning multiple values (actually a tuple)
def min_max(lst):
    return min(lst), max(lst)

lo, hi = min_max([4, 9, 1, 7])
print(lo, hi)

# *args and **kwargs
def total(*args):
    return sum(args)

def show(**kwargs):
    for k, v in kwargs.items():
        print(k, "=", v)

print(total(1, 2, 3, 4))
show(a=1, b=2)

# A function with no return gives None
def nothing():
    pass
print(nothing())
```

```python
# Scope: local vs global
count = 0

def increment():
    global count         # needed to MODIFY a global variable
    count += 1

increment(); increment()
print(count)             # 2

# Lambda: small anonymous function
square = lambda x: x * x
print(square(6))
print(sorted(["pear", "fig", "banana"], key=lambda w: len(w)))

# Recursion
def factorial(n):
    return 1 if n <= 1 else n * factorial(n - 1)

print(factorial(5))      # 120

# Functions are objects: can be passed around
def apply(fn, value):
    return fn(value)
print(apply(str.upper, "hi"))
```

## 16. Comprehensions
Compact way to build lists, sets and dicts: `[expression for item in iterable if condition]`

```python
squares = [x * x for x in range(1, 6)]
evens = [x for x in range(20) if x % 2 == 0]
labels = ["even" if x % 2 == 0 else "odd" for x in range(5)]
pairs = [(i, j) for i in range(1, 3) for j in range(1, 3)]
flat = [x for row in [[1, 2], [3, 4], [5]] for x in row]

print(squares)
print(evens)
print(labels)
print(pairs)
print(flat)

print({x % 3 for x in range(10)})                 # set comprehension
print({w: len(w) for w in ["hi", "python"]})      # dict comprehension
print(sum(x * x for x in range(4)))               # generator expression (no brackets)
```

## 17. Exception handling
Common exceptions: `ValueError`, `TypeError`, `ZeroDivisionError`, `IndexError`, `KeyError`, `FileNotFoundError`, `NameError`.

```python
def safe_divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("Cannot divide by zero")
        return None
    except TypeError as e:
        print("Bad types:", e)
        return None
    else:
        print("No error happened")     # runs only if try succeeded
        return result
    finally:
        print("finally always runs")   # cleanup code

print(safe_divide(10, 2))
print(safe_divide(10, 0))
print(safe_divide(10, "a"))

# Raising your own error
def set_age(age):
    if age < 0:
        raise ValueError("age cannot be negative")
    return age

try:
    set_age(-5)
except ValueError as e:
    print("Caught:", e)

# Catch several types together
try:
    int("abc")
except (ValueError, TypeError) as e:
    print("Conversion problem:", e)
```

## 18. File handling
Modes: `"r"` read, `"w"` write (overwrites), `"a"` append, `"r+"` read/write. Always prefer `with` so the file is closed automatically.

```python
import os

filename = "demo_file.txt"

with open(filename, "w") as f:
    f.write("Line 1\n")
    f.write("Line 2\n")
    f.writelines(["Line 3\n", "Line 4\n"])

with open(filename, "a") as f:
    f.write("Line 5\n")

with open(filename, "r") as f:
    print(f.read())               # whole file as one string

with open(filename) as f:
    for line in f:                # memory friendly, line by line
        print(line.strip())

with open(filename) as f:
    lines = f.readlines()         # list of lines
print(len(lines))

os.remove(filename)               # cleanup so no file is left behind
print(os.path.exists(filename))   # False
```

## 19. Classes & objects (OOP basics)

```python
class Student:
    school = "ABC School"                 # class attribute (shared)

    def __init__(self, name, marks):      # constructor
        self.name = name                  # instance attributes
        self.marks = marks

    def is_pass(self):                    # instance method
        return self.marks >= 40

    def __str__(self):                    # used by print()
        return f"Student({self.name}, {self.marks})"


s1 = Student("Ravi", 55)
s2 = Student("Mina", 30)
print(s1, s1.is_pass(), s2.is_pass(), Student.school)


# Inheritance + method overriding
class Person:
    def __init__(self, name):
        self.name = name

    def speak(self):
        return f"{self.name} makes a sound"


class Dog(Person):
    def __init__(self, name, breed):
        super().__init__(name)            # call parent constructor
        self.breed = breed

    def speak(self):                      # override
        return f"{self.name} the {self.breed} barks"


print(Person("Tom").speak())
print(Dog("Rex", "Labrador").speak())
```

## 21. Common exam traps ⚠️

```python
# 1) Mutable default argument
def bad(item, bucket=[]):
    bucket.append(item)
    return bucket
print(bad(1), bad(2))        # [1, 2] [1, 2]  <- shared list!

def good(item, bucket=None):
    if bucket is None:
        bucket = []
    bucket.append(item)
    return bucket
print(good(1), good(2))      # [1] [2]

# 2) == vs is
a = [1, 2]; b = [1, 2]
print(a == b, a is b)        # True False

# 3) Float precision
print(0.1 + 0.2 == 0.3)      # False
print(round(0.1 + 0.2, 2) == 0.3)   # True

# 4) / vs //
print(7 / 2, 7 // 2)         # 3.5 3

# 5) Shallow vs deep copy
import copy
orig = [[1, 2], [3, 4]]
shallow = orig.copy()
deep = copy.deepcopy(orig)
orig[0][0] = 99
print(shallow[0][0], deep[0][0])    # 99 1

# 6) list.sort() returns None
x = [3, 1, 2].sort()
print(x)                     # None

# 7) Strings are immutable
s = "hello"
s = s.upper()                # creates a new string
print(s)

# 8) Modifying a list while iterating over it
nums = [1, 2, 3, 4, 5, 6]
nums = [n for n in nums if n % 2 != 0]   # build a new list instead
print(nums)

# 9) Integer division of negatives
print(-7 // 2, -7 % 2)       # -4 1

# 10) Empty set vs dict
print(type({}), type(set()))
```

## 22. Practice questions with solutions
Try each one yourself first, then compare with the solution cell.

```python
# Q1. Swap two numbers without a third variable.
a, b = 10, 20
a, b = b, a
print(a, b)

# Q2. Check even or odd.
def even_odd(n):
    return "even" if n % 2 == 0 else "odd"
print(even_odd(7), even_odd(10))

# Q3. Largest of three numbers.
def largest(a, b, c):
    return max(a, b, c)
print(largest(4, 9, 2))

# Q4. Factorial.
def fact(n):
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result
print(fact(6))

# Q5. Palindrome check (ignore case and spaces).
def is_palindrome(s):
    cleaned = s.replace(" ", "").lower()
    return cleaned == cleaned[::-1]
print(is_palindrome("Nurses Run"), is_palindrome("python"))
```

```python
# Q6. First n Fibonacci numbers.
def fib(n):
    seq = []
    a, b = 0, 1
    for _ in range(n):
        seq.append(a)
        a, b = b, a + b
    return seq
print(fib(10))

# Q7. Prime check.
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True
print([n for n in range(1, 30) if is_prime(n)])

# Q8. Count vowels in a string.
def count_vowels(s):
    return sum(1 for ch in s.lower() if ch in "aeiou")
print(count_vowels("Programming Language"))

# Q9. Reverse the words in a sentence.
def reverse_words(sentence):
    return " ".join(sentence.split()[::-1])
print(reverse_words("python is fun"))

# Q10. Remove duplicates but keep the original order.
def unique(lst):
    seen = set()
    result = []
    for x in lst:
        if x not in seen:
            seen.add(x)
            result.append(x)
    return result
print(unique([3, 1, 3, 2, 1, 4]))
```

```python
# Q11. Are two strings anagrams?
def is_anagram(a, b):
    return sorted(a.lower()) == sorted(b.lower())
print(is_anagram("Listen", "Silent"), is_anagram("abc", "abd"))

# Q12. Word frequency.
def word_freq(text):
    freq = {}
    for w in text.lower().split():
        freq[w] = freq.get(w, 0) + 1
    return freq
print(word_freq("the cat and the hat and the bat"))

# Q13. Second largest element (distinct).
def second_largest(lst):
    unique_sorted = sorted(set(lst), reverse=True)
    return unique_sorted[1] if len(unique_sorted) > 1 else None
print(second_largest([5, 9, 2, 9, 7]))

# Q14. Sum of digits.
def digit_sum(n):
    return sum(int(d) for d in str(abs(n)))
print(digit_sum(12345))

# Q15. FizzBuzz 1..15.
for i in range(1, 16):
    if i % 15 == 0:
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)

# Q16. Transpose a matrix.
m = [[1, 2, 3], [4, 5, 6]]
print([list(row) for row in zip(*m)])

# Q17. Count characters that appear more than once.
from collections import Counter
print([ch for ch, n in Counter("programming").items() if n > 1])
```
